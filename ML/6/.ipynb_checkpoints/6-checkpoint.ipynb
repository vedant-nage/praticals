```python
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import dendrogram, fcluster, linkage
from sklearn.cluster import KMeans
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

import matplotlib.pyplot as plt

# Load the dataset from the notebook's current folder.
# Try reading with utf-8, fallback to latin1 if UnicodeDecodeError occurs
try:
    df = pd.read_csv(Path("sales.csv"))
except UnicodeDecodeError:
    df = pd.read_csv(Path("sales.csv"), encoding="latin1")

# Cluster on numeric features, excluding likely row identifiers.
features = df.select_dtypes(include="number").copy()
features = features.loc[:, features.nunique(dropna=True) > 1]
id_columns = [
    col for col in features.columns
    if col.lower() in {"id", "index"} or col.lower().endswith("_id")
]
features = features.drop(columns=id_columns, errors="ignore")

if features.empty or len(df) < 3:
    raise ValueError("Need at least one usable numeric feature and 3 rows to cluster.")

# Impute missing values and scale features.
X = SimpleImputer(strategy="median").fit_transform(features)
X = StandardScaler().fit_transform(X)

# K-Means elbow method.
max_k = min(10, len(df) - 1)
k_values = list(range(1, max_k + 1))
inertias = [
    KMeans(n_clusters=k, random_state=42, n_init=10).fit(X).inertia_
    for k in k_values
]

plt.figure(figsize=(7, 4))
plt.plot(k_values, inertias, marker="o")
plt.xticks(k_values)
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia")
plt.title("K-Means Elbow Method")
plt.grid(True)
plt.show()

# Estimate the elbow as the point farthest from the line joining the endpoints.
if len(k_values) >= 3:
    points = np.column_stack((k_values, inertias)).astype(float)
    start, end = points[0], points[-1]
    distances = np.abs(
        np.cross(end - start, points - start) / np.linalg.norm(end - start)
    )
    optimal_k = k_values[int(np.argmax(distances))]
else:
    optimal_k = 2

print(f"Estimated number of clusters: {optimal_k}")

# Fit K-Means with the estimated cluster count.
df["kmeans_cluster"] = KMeans(
    n_clusters=optimal_k, random_state=42, n_init=10
).fit_predict(X)

# Hierarchical clustering (Ward linkage). Limit the sample for performance.
max_hierarchical_rows = 2000
rng = np.random.default_rng(42)
sample_idx = (
    np.arange(len(df))
    if len(df) <= max_hierarchical_rows
    else np.sort(rng.choice(len(df), max_hierarchical_rows, replace=False))
)
X_hier = X[sample_idx]

Z = linkage(X_hier, method="ward")

plt.figure(figsize=(10, 5))
dendrogram(Z, truncate_mode="lastp", p=30)
plt.title("Hierarchical Clustering Dendrogram (Ward)")
plt.xlabel("Cluster")
plt.ylabel("Distance")
plt.show()

hierarchical_labels = fcluster(Z, t=optimal_k, criterion="maxclust")
df["hierarchical_cluster"] = pd.Series(pd.NA, index=df.index, dtype="Int64")
df.loc[sample_idx, "hierarchical_cluster"] = hierarchical_labels

print("K-Means cluster sizes:")
print(df["kmeans_cluster"].value_counts().sort_index())
print(f"\nHierarchical clustering assigned labels to {len(sample_idx)} rows.")
print(df.head())
```
