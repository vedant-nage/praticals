```python
def gradient_descent(gradient, start, learning_rate=0.1, tolerance=1e-8, max_iterations=10_000):
    x = float(start)

    for _ in range(max_iterations):
        step = learning_rate * gradient(x)
        x -= step
        if abs(step) < tolerance:
            break

    return x


# Minimize y = (x + 3)^2, whose derivative is 2(x + 3)
minimum_x = gradient_descent(lambda x: 2 * (x + 3), start=2)
minimum_y = (minimum_x + 3) ** 2

print(f"Local minimum: x = {minimum_x:.8f}, y = {minimum_y:.8f}")
```
