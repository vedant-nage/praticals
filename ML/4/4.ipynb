```python
import matplotlib.pyplot as plt

# Function and its derivative
def f(x):
    return (x + 3)**2

def df(x):
    return 2 * (x + 3)

# Initialize parameters
x_old = 2
alpha = 0.1
precision = 0.00001
max_iters = 100

# Lists for visualization
x_list = [x_old]
y_list = [f(x_old)]

# Gradient Descent Loop
for i in range(max_iters):

    grad = df(x_old)

    # Gradient Descent formula
    x_new = x_old - alpha * grad

    x_list.append(x_new)
    y_list.append(f(x_new))

    # Stopping condition
    if abs(x_new - x_old) < precision:
        break

    x_old = x_new

# Display results
print(f"Local minimum occurs at x = {x_new:.5f}")
print(f"Minimum value of function y = {f(x_new):.5f}")

# Visualization
plt.figure(figsize=(10, 5))

x_vals = [i for i in range(-10, 5)]
y_vals = [f(i) for i in x_vals]

plt.plot(x_vals, y_vals, label='y = (x+3)^2')
plt.scatter(x_list, y_list, label='Descent Steps')

plt.title('Gradient Descent Convergence')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
